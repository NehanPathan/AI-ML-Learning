```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

page_number = 1
life_quotes = []

while True:
    url = f"https://quotes.toscrape.com/page/{page_number}"
    res = requests.get(url)
    
    if res.status_code != 200:
        break
    
    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")
    
    if not quotes:
        print("No more quotes. Stopping.")
        break
    
    # ✅ Extract here (important)
    for quote in quotes:
        if any(tag.get_text(strip=True) == "life" for tag in quote.select("a.tag")):
            text = quote.select_one("span.text").get_text(strip=True)
            author = quote.select_one("small.author").get_text(strip=True)
            life_quotes.append([text, author])
    
    page_number += 1

# After loop
data = pd.DataFrame(life_quotes, columns=["Life_Quote", "Author"])
data
```
