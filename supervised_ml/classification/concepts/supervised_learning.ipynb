```python
import pandas as pd
import seaborn as sns
```

```python
insurance_data = pd.read_csv("insurance.csv")
```

```python
X = insurance_data.drop(columns=["charges", "region"])
Y = insurance_data["charges"]

X["sex"] = X["sex"].map({"female" : 1, "male" : 0})
X["smoker"] = X["smoker"].map({"yes" : 1, "no" : 0})
```

```python
X.head()
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.2,   # 20% test data
    random_state=42  # for reproducibility
)
```

```python
#Train Model
from sklearn.linear_model import LinearRegression
```

```python
# Create model
model = LinearRegression()

# Train model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)
```

```python
#Evaluate
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(y_test, y_pred)

n = len(y_test)      # number of samples or can use X_test.shape[0] both same
p = X_test.shape[1]  # number of features

adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

print("R2:", r2)
print("Adjusted R2:", adj_r2)
```

```python
#Practice Problem House Price Prediction
house_price_data = pd.read_csv("HousePricePrediction.csv")
```

```python
#Cleaning
X = house_price_data.drop(columns=["Id","SalePrice"])
Y = house_price_data["SalePrice"]

X.isnull().sum()
# Y.isnull().sum()
```

```python
# If missing:
# Numeric → fill with mean
# Eg : df.fillna(df.mean(numeric_only=True), inplace=True)

X.fillna(X.mean(numeric_only=True), inplace=True)
Y.fillna(Y.mean(numeric_only=True), inplace=True)

# Categorical → fill with most common
# Eg : df.fillna(df.mode().iloc[0], inplace=True)

X.fillna(X.mode().iloc[0], inplace=True)

X.head()
# X.isnull().sum()
```

```python
# Convert text → numbers
# Machine learning only understands numbers.
# df = pd.get_dummies(df)
# dtype=
# pd.get_dummies(df, dtype=int)
# Ensures output is integer (0/1 instead of True/False)

# X = pd.get_dummies(X, dtype=int)
X = pd.get_dummies(X, columns=["MSZoning", "LotConfig", "BldgType", "Exterior1st"], dtype=int)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.2,   # 20% test data
    random_state=42  # for reproducibility
)
# Create model
model = LinearRegression()

# Train model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluate
r2 = r2_score(y_test, y_pred)

n = len(y_test)      # number of samples or can use X_test.shape[0] both same
p = X_test.shape[1]  # number of features

adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

print("R2:", r2)
print("Adjusted R2:", adj_r2)
```
