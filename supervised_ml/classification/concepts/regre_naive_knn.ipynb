```python
import pandas as pd
from sklearn.metrics import precision_score, accuracy_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression


# convert 3 species into 0, 1, 2
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("IRIS.csv")
df.head()
le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
```

```python
#cleaning: Not Needed as seen
# 1st KNN Approch
X = df.drop(columns=["Id", "Species"])
y = df["Species"]

# split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42
)
```

```python
# Standardization
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X)
```

```python
# Model
classifier = KNeighborsClassifier()

# Parameters
param_grid = {
    "n_neighbors": [3, 5, 7, 9]
}

# Grid Search
classifierCV = GridSearchCV(
    classifier,
    param_grid,
    cv=5,
    scoring='accuracy'
)

# Train
classifierCV.fit(X_train_scaled, y_train)

# Prediction
y_pred = classifierCV.predict(X_test_scaled)

# Metrics
print("Recall score:", recall_score(y, y_pred, average='macro'))
print("Accuracy score:", accuracy_score(y, y_pred))
print("Precision score:", precision_score(y, y_pred, average='macro'))

# Results
res = pd.DataFrame(classifierCV.cv_results_)

print(res[["param_n_neighbors", "mean_test_score"]])

print("Best Params:", classifierCV.best_params_)
```

```python
# Model
classifier = GaussianNB()

# Parameters
param_grid = {
    "var_smoothing": [1e-9, 1e-8, 1e-7]
}

# Grid Search
classifierCV = GridSearchCV(
    classifier,
    param_grid,
    cv=5,
    scoring='accuracy'
)

# Train
classifierCV.fit(X_train_scaled, y_train)

# Prediction
y_pred = classifierCV.predict(X_test_scaled)

# Metrics
print("Recall score:", recall_score(y, y_pred, average='macro'))
print("Accuracy score:", accuracy_score(y, y_pred))
print("Precision score:", precision_score(y, y_pred, average='macro'))

# CV Results
res = pd.DataFrame(classifierCV.cv_results_)

print(res[["param_var_smoothing", "mean_test_score"]])

print("Best Params:", classifierCV.best_params_)
```
