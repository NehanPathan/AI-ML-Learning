```python
df = pd.read_csv("raw_data.csv")
cleaned_data = df.copy()
cleaned_data

df.isna()

df.isna().sum()

df.dropna() #row
df.dropna(axis=1) #col

mean_avg = df["age"].mean()
cleaned_data = cleaned_data["age"].fillna(mean_avg)
cleaned_data

df.ffill()
df.bfill()
```

```python
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%" )

gender_mapping = {"Male" : "M", "Female" : "F", "Unknown" : "U"}

df2["gender"] = df2["gender"].map(gender_mapping)

df2 = df2.assign(new_income = df2["income"] * 1.1)

df2["country"] = df2["country"].replace("USA", "United States")

# df2[["name", "age", "country", "gender", "income", "tax", "new_income", "id"]]
(df2["id"], df2["income"]) = (df2["income"], df2["id"])

df2.rename(columns={"id":"income", "income": "id"}, inplace=True)
df2
```

```python
df2 = df.copy()

df2 = df2.drop_duplicates()
df2= df2.fillna(0)

df2 = df2.sort_values("income")

df2 = df2.reset_index(drop=True)

df2.to_csv("sorted_data.csv")
```
