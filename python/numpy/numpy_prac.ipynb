```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5])
print(arr1, arr1.dtype, type(arr1))

arr2 = np.zeros((2,4), dtype="int64")
print(arr2, arr2.shape)


arr2 = np.ones((2,4), dtype="int64")
print(arr2, arr2.shape)


arr2 = np.full((2,4), 100 ,dtype="int64")
print(arr2, arr2.shape)


arr2 = np.eye(3, dtype="int64")
print(arr2, arr2.shape)


arr2 = np.arange(0,10,2, dtype="int64")
print(arr2, arr2.shape)


arr2 = np.linspace(1,100,3, dtype="int64")
print(arr2, arr2.shape)
```
