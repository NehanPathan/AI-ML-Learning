```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

        return data
        
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5}
    cleaned_data = []
    unique_user = set()
    
    for user in data:

        #Cleaning our data
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        #Handle missing values
        age = user.get("age")
        if(age == None):
            user["age"] = None

        #Deduplication
        if user["name"].strip() in unique_user:
            continue

        unique_user.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
        
```

```python
data = clean_data(data)
```

```python
#get Meaningful insights from our data
def get_insights(data):
    avg_rating = 0
    poor_rating = 0
    
    #Avg. rating and % of user giving poor rating ie <3
    
    for user in data:
        if(float(user["rating"]) < 3):
            poor_rating += 1
            avg_rating += float(user["rating"])
            continue
         
        avg_rating += float(user["rating"])
           
        

    print(f"avg rating is {avg_rating / len(data)}")
    print(f"% of user giving poor rating is {(poor_rating / len(data)) * 100}%")

```

```python
get_insights(data)
```

```python
#Build recommendation based on rating
def get_recommendation(data):
    recommendation = []

    for user in data:
        curr_rec = {}
        curr_rec["name"] = user["name"]

        if float(user["rating"]) > 4:
            curr_rec["brand"] = "Apple"
        else:
            curr_rec["brand"] = "Samsung"

        recommendation.append(curr_rec)
        
    return recommendation        
            
```

```python
get_recommendation(data)
```
